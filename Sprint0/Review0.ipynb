```polyglot-notebook
class Plant
{
    public string Name {get; set; }
    public string Type {get; set; }
    public float Height {get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }


    public virtual void Blossom()
    {
        Console.WriteLine($"Растение {Name} цветет!");
    }
    
    public virtual void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"Растение {Name} выросло на {growth} см. Текущая высота: {Height} см.");
    }

}

class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, float height, string color, int petalCount) : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Blossom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wilt()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать");
    }
}

class Tree : Plant
{
    public int Age { get; set;}
    public string LeafType { get; set; }

    public Tree(string name, float height, int age, string leafType) : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Blossom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowYears(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} выросло. Новый возраст: {Age} лет");
    }
}

Flower sunflower = new Flower("Подсолнух", 70.5f, "Желтый", 30);
Tree cherry = new Tree("Вишня", 170.0f, 5, "Лиственные");


sunflower.Blossom();
cherry.Blossom();


sunflower.Grow(4.5f);
cherry.GrowYears(3);


sunflower.Wilt();

Console.WriteLine($"Цвет подсолнуха: {sunflower.Color}, лепестков: {sunflower.PetalCount}");
Console.WriteLine($"Тип листьев вишни: {cherry.LeafType}, возраст: {cherry.Age}");
```
