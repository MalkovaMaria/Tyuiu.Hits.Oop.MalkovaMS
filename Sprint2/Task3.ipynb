Описание задачи: 
Создать базовый класс PaymentMethod в C#, который будет представлять 
различные способы оплаты. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 
Требования к базовому классу PaymentMethod: 

• Атрибуты: 
1. ID способа оплаты (PaymentMethodId)
2. Название способа оплаты (MethodName)
3. Минимальная сумма (MinAmount). 

• Методы: 
1. ProcessPayment(decimal amount): метод для обработки платежа 
указанной суммы. 
2. CheckMinimumAmount(decimal amount): метод для проверки 
минимальной суммы платежа. 
3. GetPaymentDetails(): метод для получения деталей способа оплаты.
 
Требования к производным классам: 
1. ОнлайнОплата (OnlinePayment): Должен содержать дополнительные 
атрибуты, 
такие 
как 
URL платежной системы (PaymentUrl). 
Метод ProcessPayment() должен быть переопределен для включения URL 
платежной системы в процесс оплаты. 
2. БанковскийПеревод (BankTransfer): Должен содержать дополнительные 
атрибуты, 
такие 
как 
Банковские 
данные 
(BankData). 
Метод CheckMinimumAmount() должен быть переопределен для проверки 
минимальной суммы платежа с учетом банковских комиссий. 
3. Наличные (CashPayment) (если требуется третий класс): Должен содержать 
дополнительные атрибуты, такие как Место выдачи наличных 
(CashPickupPoint). Метод GetPaymentDetails() должен быть переопределен 
для отображения места выдачи наличных.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
public interface IRefundable
{
    bool Refund(decimal amount);
}

public class PaymentMethod
{
    private int _paymentMethodId;
    private string _methodName;
    private decimal _minAmount;
    protected string _currency;
    protected bool _isEnabled;
    protected int _transactionCount;

    public int PaymentMethodId
    {
        get { return _paymentMethodId; }
        set { _paymentMethodId = value; }
    }

    public string MethodName
    {
        get { return _methodName; }
        set { _methodName = value; }
    }

    public decimal MinAmount
    {
        get { return _minAmount; }
        set 
        {
            if (value >= 0) _minAmount = value; 
            else Console.WriteLine("Ошибка: сумма не может быть отрицательной!");
        }
    }

    public string Currency
    {
        get { return _currency; }
        set { _currency = value; }
    }

    public bool IsEnabled
    {
        get { return _isEnabled; }
        set { _isEnabled = value; }
    }

    public PaymentMethod(int id, string name, decimal minAmount, string currency)
    {
        this.PaymentMethodId = id;
        this.MethodName = name; 
        this.MinAmount = minAmount;
        this.Currency = currency;
        this.IsEnabled = true;
        this._transactionCount = 0;
    }

    public virtual void ProcessPayment(decimal amount)
    {
        if (!IsEnabled)
        {
            Console.WriteLine($"Метод '{MethodName}' временно отключен.");
            return;
        }

        if (CheckMinimumAmount(amount))
        {
            _transactionCount++;
            Console.WriteLine($"Платеж через {MethodName} в размере {amount} {Currency} обработан");
        }
        else
        {
            Console.WriteLine($"Платеж методом '{MethodName}' в размере {amount} {Currency} отклонен.");
        }
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        return amount >= MinAmount;
    }

    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"Детали: ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount} {Currency}, Статус: {(IsEnabled ? "Активен" : "Отключен")}");
    }

    public void Disable()
    {
        IsEnabled = false;
        Console.WriteLine($"Метод оплаты '{MethodName}' отключен.");
    }

    public int GetTransactionCount()
    {
        return _transactionCount;
    }
}

public class OnlinePayment : PaymentMethod, IRefundable
{
    private string _paymentUrl;
    private decimal _transactionFee;
    private bool _isSecure;

    public string PaymentUrl
    {
        get { return _paymentUrl; }
        set { _paymentUrl = value; }
    }

    public OnlinePayment(int id, string name, decimal minAmount, string currency, string url, decimal fee) 
        : base(id, name, minAmount, currency)
    {
        this.PaymentUrl = url;
        this._transactionFee = fee;
        this._isSecure = true;
    }

    public override void ProcessPayment(decimal amount)
    {
        base.ProcessPayment(amount); 
        if (amount >= MinAmount && IsEnabled) 
        {
            Console.WriteLine($"--> Оплата проведена по URL: {PaymentUrl} (Комиссия сервиса: {_transactionFee} {Currency})");
        }
    }

    public bool Refund(decimal amount)
    {
        Console.WriteLine($"[Онлайн] Возврат {amount} {Currency} инициирован. Средства вернутся в течение 3 дней.");
        return true;
    }

    public void EnableTwoFactorAuth()
    {
        _isSecure = true;
        Console.WriteLine("Двухфакторная аутентификация включена.");
    }
}

public class BankTransfer : PaymentMethod
{
    private decimal _bankData;
    private string _bankName;
    private int _processingTimeHours;

    public decimal BankData
    {
        get { return _bankData; }
        set { _bankData = value; }
    }

    public BankTransfer(int id, string name, decimal minAmount, string currency, decimal bankData, string bankName, int time) 
        : base(id, name, minAmount, currency)
    {
        this.BankData = bankData;
        this._bankName = bankName;
        this._processingTimeHours = time;
    }

    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount / 100 * BankData;
        decimal amountAfterCommission = amount - commission;

        if (amountAfterCommission >= MinAmount)
        {
            Console.WriteLine($"(Сумма достаточна. Комиссия {_bankName}: {commission} {Currency})");
            return true;
        }
        else
        {
            Console.WriteLine($"(Недостаточно средств. После комиссии останется {amountAfterCommission} {Currency}, а нужно {MinAmount})");
            return false;
        }
    }

    public void EstimateArrivalTime()
    {
        Console.WriteLine($"Ожидаемое время зачисления средствами {_bankName}: {_processingTimeHours} ч.");
    }
}

public class CashPayment : PaymentMethod
{
    private string _cashPickupPoint;
    private string _operatorName;
    private bool _requiresId;

    public string CashPickupPoint
    {
        get { return _cashPickupPoint; }
        set { _cashPickupPoint = value; }
    }

    public CashPayment(int id, string name, decimal minAmount, string currency, string pickupPoint, string operatorName, bool requiresId) 
        : base(id, name, minAmount, currency)
    {
        this.CashPickupPoint = pickupPoint;
        this._operatorName = operatorName;
        this._requiresId = requiresId;
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"--> Место выдачи: {CashPickupPoint}, Оператор: {_operatorName}, Требуется паспорт: {_requiresId}");
    }

    public void PrintReceipt()
    {
        Console.WriteLine($"Чек распечатан оператором {_operatorName}");
    }
}

public class CryptoPayment : OnlinePayment
{
    private string _walletAddress;
    private string _networkName;

    public CryptoPayment(int id, string name, decimal minAmount, string currency, string url, decimal fee, string wallet, string network) 
        : base(id, name, minAmount, currency, url, fee)
    {
        this._walletAddress = wallet;
        this._networkName = network;
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"--> Крипто-кошелек: {_walletAddress}, Сеть: {_networkName}");
    }

    public void VerifyBlockchain()
    {
        Console.WriteLine($"Проверка транзакции в сети {_networkName}... Подтверждено.");
    }
}

public class Customer
{
    public string Name { get; set; }

    public Customer(string name)
    {
        Name = name;
    }

    public void MakePurchase(PaymentMethod method, decimal amount)
    {
        Console.WriteLine($"\n[Покупатель {Name} пытается заплатить {amount} руб.]");
        method.GetPaymentDetails();
        method.ProcessPayment(amount);
    }

    public void RequestRefund(IRefundable refundableMethod, decimal amount)
    {
        Console.WriteLine($"\n[Покупатель {Name} запрашивает возврат]");
        refundableMethod.Refund(amount);
    }
}

BankTransfer bank = new BankTransfer(2, "Банковский перевод", 50, "RUB", 5, "Сбербанк", 24);
CashPayment cash = new CashPayment(3, "Наличные", 100, "RUB", "Газпромбанк", "Ольга", true);
OnlinePayment online = new OnlinePayment(1, "Онлайн оплата", 10, "RUB", "https://api.bank.ru/pay", 2);
CryptoPayment crypto = new CryptoPayment(4, "Крипто-оплата", 500, "USDT", "https://crypto.gate", 1, "0x123...abc", "Ethereum");

Customer buyer = new Customer("Кирилл");

Console.WriteLine("=== ТЕСТ 1: Простое наследование и новые методы ===");
bank.EstimateArrivalTime();
buyer.MakePurchase(bank, 200);
cash.PrintReceipt();
buyer.MakePurchase(cash, 150);

Console.WriteLine("\n=== ТЕСТ 2: Множественное наследование ===");
buyer.MakePurchase(online, 150);
buyer.RequestRefund(online, 50);

Console.WriteLine("\n=== ТЕСТ 3: Сложное наследование ===");
crypto.VerifyBlockchain();
buyer.MakePurchase(crypto, 1000);

Console.WriteLine("\n=== ТЕСТ 4: Статистика базового класса ===");
Console.WriteLine($"Всего успешных транзакций через Онлайн: {online.GetTransactionCount()}");
Console.WriteLine($"Всего успешных транзакций через Крипто: {crypto.GetTransactionCount()}");

online.Disable();
buyer.MakePurchase(online, 100);
```
