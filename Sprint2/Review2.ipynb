```polyglot-notebook
public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeade
{
    string[] GetTeamMembers();
}

class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public float Salary { get; set; }

    public Employee(string name, string position, float salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, Должность: {Position}, Зарплата: {Salary} руб. ");
    }
}

class Manager : Employee
{
    public string Department { get; set; }
    
    public Manager(string name, string position, float salary, string department) : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

class ProjectManager : Manager, IReportable, ITeamLeade
{
    public string ProjectName {get; set;}
    private string[] TeamMembers;

    public ProjectManager(string name, string position, float salary, string department, string projectName, string [] teamMembers) : base(name, position, salary, department)
    {
        ProjectName = projectName;
        TeamMembers = teamMembers;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Название проекта: {ProjectName}");
    }

    public string GenerateReport()
    {
        return $"Отчёт по проекту {ProjectName} сформирован";
    }

    public string[] GetTeamMembers()
    {
        return TeamMembers;
    }
    
}

Employee employee = new Employee("Кирилл", "Консультант", 50000f);
Manager manager = new Manager("Настя", "Менеджен", 58000f, "Продажи");
ProjectManager projectManager = new ProjectManager("Алексей", "Проектный менеджер", 670000f, "Маркетинг", "Тикток хаус",  new string[] { "Андрей", "Юлия", "Маргарита", "Мирослав" });

employee.PrintInfo();
Console.WriteLine();

manager.PrintInfo();
Console.WriteLine();

projectManager.PrintInfo();
Console.WriteLine();

Console.WriteLine(projectManager.GenerateReport());

Console.WriteLine("Команда проекта:");
foreach (string member in projectManager.GetTeamMembers())
{
    Console.WriteLine(member);
}
```
