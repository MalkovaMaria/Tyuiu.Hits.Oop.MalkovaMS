```polyglot-notebook
using System;
using System.Collections.Generic;
using System.Linq;

public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Manufacturer { get; set; }

    protected Component(string name, decimal price, string manufacturer)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, Производитель: {Manufacturer}, Цена: {Price} руб.");
    }

    public override string ToString()
    {
        return $"[{GetType().Name}] {Name} от {Manufacturer} - {Price} руб.";
    }
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double FrequencyGHz { get; set; }

    public CPU(string name, decimal price, string manufacturer, int cores, double frequencyGHz)
        : base(name, price, manufacturer)
    {
        Cores = cores;
        FrequencyGHz = frequencyGHz;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  -> Ядра: {Cores}, Частота: {FrequencyGHz} ГГц");
    }

    public void Install()
    {
        Console.WriteLine($"[УСТАНОВКА] Процессор {Name} установлен в сокет материнской платы.");
    }

    public static bool operator ==(CPU left, CPU right)
    {
        if (ReferenceEquals(left, null) && ReferenceEquals(right, null))
            return true;
        if (ReferenceEquals(left, null) || ReferenceEquals(right, null))
            return false;
        return left.Name == right.Name 
            && left.Manufacturer == right.Manufacturer 
            && left.Price == right.Price 
            && left.Cores == right.Cores 
            && left.FrequencyGHz == right.FrequencyGHz;
    }

    public static bool operator !=(CPU left, CPU right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj) => obj is CPU other && Name == other.Name && Manufacturer == other.Manufacturer && Price == other.Price && Cores == other.Cores && FrequencyGHz == other.FrequencyGHz;

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Price, Cores, FrequencyGHz);
    }
}

public class RAM : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public string Type { get; set; }

    public RAM(string name, decimal price, string manufacturer, int capacityGB, string type)
        : base(name, price, manufacturer)
    {
        CapacityGB = capacityGB;
        Type = type;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  -> Объем: {CapacityGB} ГБ, Тип: {Type}");
    }

    public void Install()
    {
        Console.WriteLine($"[УСТАНОВКА] Оперативная память {Name} ({CapacityGB} ГБ) установлена в слот DIMM.");
    }

    public static bool operator ==(RAM left, RAM right)
    {
        if (ReferenceEquals(left, null) && ReferenceEquals(right, null))
            return true;
        if (ReferenceEquals(left, null) || ReferenceEquals(right, null))
            return false;
        return left.Name == right.Name 
            && left.Manufacturer == right.Manufacturer 
            && left.Price == right.Price 
            && left.CapacityGB == right.CapacityGB 
            && left.Type == right.Type;
    }

    public static bool operator !=(RAM left, RAM right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj) => obj is RAM other && Name == other.Name && Manufacturer == other.Manufacturer && Price == other.Price && CapacityGB == other.CapacityGB && Type == other.Type;

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Price, CapacityGB, Type);
    }
}

public class GPU : Component, IInstallable
{
    public int MemoryGB { get; set; }

    public GPU(string name, decimal price, string manufacturer, int memoryGB)
        : base(name, price, manufacturer)
    {
        MemoryGB = memoryGB;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  -> Память: {MemoryGB} ГБ");
    }

    public void Install()
    {
        Console.WriteLine($"[УСТАНОВКА] Видеокарта {Name} установлена в слот PCIe.");
    }

    public static bool operator ==(GPU left, GPU right)
    {
        if (ReferenceEquals(left, null) && ReferenceEquals(right, null))
            return true;
        if (ReferenceEquals(left, null) || ReferenceEquals(right, null))
            return false;
        return left.Name == right.Name 
            && left.Manufacturer == right.Manufacturer 
            && left.Price == right.Price 
            && left.MemoryGB == right.MemoryGB;
    }

    public static bool operator !=(GPU left, GPU right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj) => obj is GPU other && Name == other.Name && Manufacturer == other.Manufacturer && Price == other.Price && MemoryGB == other.MemoryGB;

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Price, MemoryGB);
    }
}

public class HDD : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public int RPM { get; set; }

    public HDD(string name, decimal price, string manufacturer, int capacityGB, int rpm)
        : base(name, price, manufacturer)
    {
        CapacityGB = capacityGB;
        RPM = rpm;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"  -> Объем: {CapacityGB} ГБ, Скорость: {RPM} об/мин");
    }

    public void Install()
    {
        Console.WriteLine($"[УСТАНОВКА] Накопитель {Name} ({CapacityGB} ГБ) подключен.");
    }

    public static bool operator ==(HDD left, HDD right)
    {
        if (ReferenceEquals(left, null) && ReferenceEquals(right, null))
            return true;
        if (ReferenceEquals(left, null) || ReferenceEquals(right, null))
            return false;
        return left.Name == right.Name 
            && left.Manufacturer == right.Manufacturer 
            && left.Price == right.Price 
            && left.CapacityGB == right.CapacityGB 
            && left.RPM == right.RPM;
    }

    public static bool operator !=(HDD left, HDD right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj) => obj is HDD other && Name == other.Name && Manufacturer == other.Manufacturer && Price == other.Price && CapacityGB == other.CapacityGB && RPM == other.RPM;

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, Price, CapacityGB, RPM);
    }
}

public class Computer
{
    private List<Component> _components = new List<Component>();
    public string Model { get; set; }

    public Computer(string model)
    {
        Model = model;
    }

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        Console.WriteLine($"[КОМПЬЮТЕР] Компонент {component.Name} добавлен в систему '{Model}'.");
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Remove(component))
        {
            Console.WriteLine($"[КОМПЬЮТЕР] Компонент {component.Name} удален из системы '{Model}'.");
        }
        else
        {
            Console.WriteLine($"[ОШИБКА] Компонент {component.Name} не найден в системе '{Model}'.");
        }
    }

    public decimal GetTotalPrice()
    {
        return _components.Sum(c => c.Price);
    }

    public void DisplayInfo()
    {
        Console.WriteLine($"\n=== Конфигурация компьютера '{Model}' ===");
        if (_components.Count == 0)
        {
            Console.WriteLine("Компьютер пуст.");
        }
        else
        {
            foreach (var component in _components)
            {
                component.GetInfo();
            }
            Console.WriteLine($"Общая стоимость: {GetTotalPrice()} руб.");
        }
    }
}

public static class ComputerFactory
{
    public static Computer CreateComputer<T1, T2, T3, T4>(string model, T1 cpu, T2 ram, T3 gpu, T4 hdd)
    where T1 : Component, IInstallable
    where T2 : Component, IInstallable
    where T3 : Component, IInstallable
    where T4 : Component, IInstallable
    {
        Computer computer = new Computer(model);
        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(gpu);
        computer.AddComponent(hdd);
        return computer;
    }

    public static Computer CreateGamingPC()
    {
        var cpu = new CPU("Intel Core i9-13900K", 55000, "Intel", 24, 5.8);
        var ram = new RAM("Kingston Fury Beast", 18000, "Kingston", 32, "DDR5");
        var gpu = new GPU("NVIDIA RTX 4090", 180000, "NVIDIA", 24);
        var hdd = new HDD("Samsung 990 Pro", 25000, "Samsung", 2000, 0);

        Computer pc = new Computer("Игровой ПК");
        pc.AddComponent(cpu);
        pc.AddComponent(ram);
        pc.AddComponent(gpu);
        pc.AddComponent(hdd);
        return pc;
    }

    public static Computer CreateOfficePC()
    {
        var cpu = new CPU("Intel Core i3-12100", 12000, "Intel", 4, 4.3);
        var ram = new RAM("Corsair Vengeance", 5000, "Corsair", 16, "DDR4");
        var hdd = new HDD("WD Blue", 4000, "Western Digital", 1000, 7200);

        Computer pc = new Computer("Офисный ПК");
        pc.AddComponent(cpu);
        pc.AddComponent(ram);
        pc.AddComponent(hdd);
        return pc;
    }
}

CPU cpu1 = new CPU("Intel Core i7-12700K", 35000, "Intel", 12, 5.0);
CPU cpu2 = new CPU("AMD Ryzen 9 5900X", 38000, "AMD", 12, 4.8);
RAM ram1 = new RAM("Corsair Vengeance LPX", 15000, "Corsair", 32, "DDR4");
GPU gpu1 = new GPU("NVIDIA GeForce RTX 3080", 90000, "NVIDIA", 10);
HDD hdd1 = new HDD("WD Blue", 6000, "Western Digital", 2000, 7200);

Console.WriteLine("--- Перегрузка ToString() ---");
Console.WriteLine(cpu1.ToString());
Console.WriteLine(ram1.ToString());
Console.WriteLine(gpu1.ToString());
Console.WriteLine(hdd1.ToString());

Console.WriteLine("\n--- Перекрытие GetInfo() ---");
cpu1.GetInfo();
ram1.GetInfo();
gpu1.GetInfo();
hdd1.GetInfo();

Console.WriteLine("\n--- Перегрузка операторов == и != ---");
Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}");
Console.WriteLine($"cpu1 != cpu2: {cpu1 != cpu2}");

CPU cpu1Copy = new CPU("Intel Core i7-12700K", 35000, "Intel", 12, 5.0);
Console.WriteLine($"cpu1 == cpu1Copy (одинаковые): {cpu1 == cpu1Copy}");

Console.WriteLine("\n--- Использование Generic методов ---");
Computer myPC = new Computer("Мой ПК");
myPC.AddComponent(cpu1);
myPC.AddComponent(ram1);
myPC.AddComponent(gpu1);
myPC.AddComponent(hdd1);

myPC.DisplayInfo();

Console.WriteLine("\n--- Удаление компонента ---");
myPC.RemoveComponent(hdd1);

Console.WriteLine("\n--- Создание компьютеров через Factory ---");
Computer gamingPC = ComputerFactory.CreateGamingPC();
gamingPC.DisplayInfo();
Console.WriteLine("\n");
Computer officePC = ComputerFactory.CreateOfficePC();
officePC.DisplayInfo();

Console.WriteLine("\n");
Computer customPC = ComputerFactory.CreateComputer("Кастомный ПК", cpu2, ram1, gpu1, hdd1);
customPC.DisplayInfo();
```
