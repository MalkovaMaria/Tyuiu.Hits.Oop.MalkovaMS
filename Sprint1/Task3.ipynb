Описание задачи: 
Создать базовый класс PaymentMethod в C#, который будет представлять 
различные способы оплаты. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 
Требования к базовому классу PaymentMethod: 

• Атрибуты: 
1. ID способа оплаты (PaymentMethodId)
2. Название способа оплаты (MethodName)
3. Минимальная сумма (MinAmount). 

• Методы: 
1. ProcessPayment(decimal amount): метод для обработки платежа 
указанной суммы. 
2. CheckMinimumAmount(decimal amount): метод для проверки 
минимальной суммы платежа. 
3. GetPaymentDetails(): метод для получения деталей способа оплаты.
 
Требования к производным классам: 
1. ОнлайнОплата (OnlinePayment): Должен содержать дополнительные 
атрибуты, 
такие 
как 
URL платежной системы (PaymentUrl). 
Метод ProcessPayment() должен быть переопределен для включения URL 
платежной системы в процесс оплаты. 
2. БанковскийПеревод (BankTransfer): Должен содержать дополнительные 
атрибуты, 
такие 
как 
Банковские 
данные 
(BankData). 
Метод CheckMinimumAmount() должен быть переопределен для проверки 
минимальной суммы платежа с учетом банковских комиссий. 
3. Наличные (CashPayment) (если требуется третий класс): Должен содержать 
дополнительные атрибуты, такие как Место выдачи наличных 
(CashPickupPoint). Метод GetPaymentDetails() должен быть переопределен 
для отображения места выдачи наличных.

```
public class PaymentMethod
{
    private int _paymentMethodId;
    private string _methodName;
    private decimal _minAmount;

    public int PaymentMethodId
    {
        get { return _paymentMethodId; }
        set { _paymentMethodId = value; }
    }

    public string MethodName
    {
        get { return _methodName; }
        set { _methodName = value; }
    }

    public decimal MinAmount
    {
        get { return _minAmount; }
        set 
        {
            if (value >= 0) 
                _minAmount = value; 
            else 
                Console.WriteLine("Ошибка: сумма не может быть отрицательной!");
        }
    }

    public PaymentMethod(int id, string name, decimal minAmount)
    {
        this.PaymentMethodId = id;
        this.MethodName = name; 
        this.MinAmount = minAmount;
    }

    public virtual void ProcessPayment(decimal amount)
    {
        if (CheckMinimumAmount(amount))
        {
            Console.WriteLine($"Платеж через {MethodName} в размере {amount} руб. обработан");
        }
        else
        {
            Console.WriteLine($"Платеж методом '{MethodName}' в размере {amount} руб. отклонен.");
        }
    }

    public virtual bool CheckMinimumAmount(decimal amount)
    {
        if (amount >= MinAmount)
        {
            return true;
        }
        else
        {
            Console.WriteLine($"Минимальная сумма не набрана. Нужно минимум {MinAmount} руб.");
            return false;
        }
    }

    public virtual void GetPaymentDetails()
    {
        Console.WriteLine($"Детали: ID: {PaymentMethodId}, Способ: {MethodName}, Мин. сумма: {MinAmount} руб.");
    }
}

public class OnlinePayment : PaymentMethod
{
    private string _paymentUrl;
    public string PaymentUrl
    {
        get { return _paymentUrl; }
        set { _paymentUrl = value; }
    }

    public OnlinePayment(int id, string name, decimal minAmount, string url) : base(id, name, minAmount)
    {
        this.PaymentUrl = url;
    }

    public override void ProcessPayment(decimal amount)
    {
        base.ProcessPayment(amount); 
        if (amount >= MinAmount) 
        {
            Console.WriteLine($"--> Оплата проведена по URL: {PaymentUrl}");
        }
    }
}

public class BankTransfer : PaymentMethod
{
    private decimal _bankData;
    public decimal BankData
    {
        get { return _bankData; }
        set { _bankData = value; }
    }

    public BankTransfer(int id, string name, decimal minAmount, decimal bankData) : base(id, name, minAmount)
    {
        this.BankData = bankData;
    }

    public override bool CheckMinimumAmount(decimal amount)
    {
        decimal commission = amount / 100 * BankData;
        decimal amountAfterCommission = amount - commission;

        if (amountAfterCommission >= MinAmount)
        {
            Console.WriteLine($"(Сумма достаточна. Комиссия банка: {commission} руб.)");
            return true;
        }
        else
        {
            Console.WriteLine($"(Недостаточно средств. После комиссии {commission} руб. останется {amountAfterCommission} руб., а нужно {MinAmount})");
            return false;
        }
    }
}

public class CashPayment : PaymentMethod
{
    private string _cashPickupPoint;
    public string CashPickupPoint
    {
        get { return _cashPickupPoint; }
        set { _cashPickupPoint = value; }
    }

    public CashPayment(int id, string name, decimal minAmount, string pickupPoint) : base(id, name, minAmount)
    {
        this.CashPickupPoint = pickupPoint;
    }

    public override void GetPaymentDetails()
    {
        base.GetPaymentDetails();
        Console.WriteLine($"--> Место выдачи наличных: {CashPickupPoint}");
    }
}


public class Customer
{
    public string Name { get; set; }

    public Customer(string name)
    {
        Name = name;
    }

    public void MakePurchase(PaymentMethod method, decimal amount)
    {
        Console.WriteLine($"\n[Покупатель {Name} пытается заплатить {amount} руб.]");
        method.GetPaymentDetails();
        method.ProcessPayment(amount);
    }
}


OnlinePayment online = new OnlinePayment(1, "Онлайн оплата", 10, "https://api.bank.ru/pay");
BankTransfer bank = new BankTransfer(2, "Банковский перевод", 50, 10);
CashPayment cash = new CashPayment(3, "Наличные", 100, "Газпромбанк");

Customer buyer = new Customer("Кирилл");

buyer.MakePurchase(online, 150);
buyer.MakePurchase(bank, 150);
buyer.MakePurchase(cash, 80);

Console.WriteLine($"\nURL для онлайн оплаты: {online.PaymentUrl}");
Console.WriteLine($"Место выдачи наличных: {cash.CashPickupPoint}");
```
